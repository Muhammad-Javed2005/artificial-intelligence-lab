## Exercise 1: Python Lambda Functions
### Task I: Square and Cube Numbers using Lambda

```python
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, numbers))
cubes = list(map(lambda x: x**3, numbers))

print("Original List:", numbers)
print("Squared List:", squares)
print("Cubed List:", cubes)
```

### Task II: Check String Start Character using Lambda

```python
starts_with = lambda string, char: string.lower().startswith(char.lower())
test_str = "Python"
test_char = "P"

print(f"Does '{test_str}' start with '{test_char}'? -> {starts_with(test_str, test_char)}")
```

### Task III: Extract Year, Month, Date, and Time using Lambda

```python
import datetime

now = datetime.datetime.now()
get_year = lambda dt: dt.year
get_month = lambda dt: dt.month
get_day = lambda dt: dt.day
get_time = lambda dt: dt.strftime("%H:%M:%S")

print("Current Datetime:", now)
print("Year:", get_year(now))
print("Month:", get_month(now))
print("Day:", get_day(now))
print("Time:", get_time(now))
```

## Exercise 2: File Handling (I/O)
### Task I: Store City Details in File

```python
num_cities = int(input("How many cities do you want to enter? "))

with open("cities.txt", "w") as file:
    for i in range(num_cities):
        print(f"\nEntering details for City {i + 1}:")
        name = input("City Name: ")
        population = input("Population: ")
        mayor = input("Mayor Name: ")
        file.write(f"City: {name}, Population: {population}, Mayor: {mayor}\n")

print("\nCity data successfully saved to 'cities.txt'.")
```

### Task II: Create `student.txt` and Append Message

## Exercise 3: Python Built-in Modules
### Module Imports, Aliasing, `time`, `glob`, and `random` Modules

```python
import math
import math as maths
from math import sqrt as squareroot
import time
import glob
import random

# Math Aliasing
print("math.sqrt(9) =", math.sqrt(9))
print("maths.sqrt(16) =", maths.sqrt(16))
print("squareroot(25) =", squareroot(25))

# Time Module
print("\nCurrent Time:", time.ctime())
time.sleep(1)

# Glob Module
print("\nPython Files in Directory:", glob.glob("*.py"))

# Random Module
print("\nRandom Integer (1 to 10):", random.randint(1, 10))
x = [1, 2, 3, 4, 5]
random.shuffle(x)
print("Shuffled List:", x)
```
