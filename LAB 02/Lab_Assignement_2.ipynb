```python
# Loop Example 2: Count Positive and Negative numbers
n = int(input("How many numbers do you want to enter? "))
pos = neg = zero = 0

for i in range(n):
    val = float(input(f"Enter number {i+1}: "))
    if val > 0:
        pos += 1
    elif val < 0:
        neg += 1
    else:
        zero += 1

print(f"Positive Numbers: {pos}")
print(f"Negative Numbers: {neg}")
print(f"Zeros: {zero}")
```

#### Question 2: What's the difference between a while loop and a for loop?
**Answer:**
A `for` loop is used to iterate over a fixed sequence or collection when the number of iterations is known in advance. On the other hand, a `while` loop repeatedly executes a block of code as long as a specified condition remains `True`. It is ideal when the exact number of iterations is unknown before execution.

```python
# Example demonstrating For Loop vs While Loop
print("For Loop Execution:")
for item in ["A", "B", "C"]:
    print(item)

print("\nWhile Loop Execution:")
count = 1
while count <= 3:
    print(f"Count: {count}")
    count += 1
```

#### Question 3: Can you sum numbers in a while loop?
**Answer:**
Yes, you can easily sum numbers inside a `while` loop by using an accumulator variable. You initialize a sum variable to zero before the loop and add the current number to it during each iteration until the condition becomes false.

```python
# Summing numbers from 1 to 5 using a while loop
total = 0
num = 1
while num <= 5:
    total += num
    num += 1
print("Sum of numbers 1 to 5 using while loop =", total)
```

#### Question 4: Can a for loop be used inside a while loop?
**Answer:**
Yes, a `for` loop can easily be nested inside a `while` loop. In this structure, the outer `while` loop controls the overall condition-based iterations, and for each iteration of the `while` loop, the inner `for` loop runs completely through its sequence.

```python
# Nested For Loop inside While Loop
outer = 1
while outer <= 2:
    print(f"Outer While Iteration {outer}:")
    for inner in [10, 20]:
        print(f"   Inner For Value: {inner}")
    outer += 1
```
