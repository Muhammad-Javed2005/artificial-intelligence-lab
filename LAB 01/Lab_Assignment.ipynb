## Task 01: Swap Four Variables
**Question:** Write a program to input four variables and swap their values such that `a -> d`, `b -> c`, `c -> b`, and `d -> a`.

```python
# Lab Task 01 - Question 1: Swap Four Variables

a = int(input("Enter value for a: "))
b = int(input("Enter value for b: "))
c = int(input("Enter value for c: "))
d = int(input("Enter value for d: "))

print("\nBefore Swapping")
print(f"a = {a}, b = {b}, c = {c}, d = {d}")

# Swapping variables
a, b, c, d = d, c, b, a

print("\nAfter Swapping")
print(f"a = {a}, b = {b}, c = {c}, d = {d}")
```

## Task 02: Temperature Converter (Celsius to Fahrenheit)
**Question:** Write a program that converts temperature from Celsius to Fahrenheit.

```python
# Lab Task 01 - Question 2: Celsius to Fahrenheit

celsius = float(input("Enter temperature in Celsius: "))

# Formula conversion
fahrenheit = (celsius * 9/5) + 32

print(f"{celsius}°C is equal to {fahrenheit}°F")
```

## Task 03: Count Strings
**Question:** Write a Python program to count the number of strings where the string length is 2 or more and the first and last character are same from a given list of strings.

```python
# Lab Task 03 - Question 2: Count Strings

strings = ['abc', 'xyz', 'aba', '1221']
count = 0

for word in strings:
    if len(word) >= 2 and word[0] == word[-1]:
        count += 1

print("List:", strings)
print("Expected Result:", count)
```

## Task 04: Dictionary Methods & Concatenation
**Question:** Demonstrate various built-in methods of Python Dictionaries (like `get`, `keys`, `values`, `items`, `update`, `pop`, `copy`, `clear`) and concatenate multiple dictionaries into a single dictionary.

```python
# Lab Task 04 - Topic: Dictionaries

print("Dictionary Methods")

# Show all dictionary methods
print(dir(dict))

print("\nHelp for dict.get()")
help(dict.get)

print("\nHelp for dict.update() ")
help(dict.update)

# Create a dictionary
student = {
    "name": "Ali",
    "age": 20
}

print("\nOriginal Dictionary:")
print(student)

# get() method
print("\nUsing get()")
print(student.get("name"))

# keys() method
print("\nUsing keys()")
print(student.keys())

# values() method
print("\nUsing values()")
print(student.values())

# items() method
print("\nUsing items()")
print(student.items())

# Add a new key
student["city"] = "Karachi"
print("\nAfter Adding City:")
print(student)

# update() method
student.update({"age": 21})
print("\nAfter update():")
print(student)

# pop() method
student.pop("city")
print("\nAfter pop():")
print(student)

# copy() method
new_student = student.copy()
print("\nCopied Dictionary:")
print(new_student)

# clear() method
temp = {"A": 1, "B": 2}
temp.clear()
print("\nAfter clear():")
print(temp)

print("\n===== Concatenate Dictionaries =====")

dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

result = {}
result.update(dic1)
result.update(dic2)
result.update(dic3)

print("Dictionary 1:", dic1)
print("Dictionary 2:", dic2)
print("Dictionary 3:", dic3)

print("\nFinal Dictionary:")
print(result)
```

## Task 05: List Comprehensions
**Question:**
1. Convert strings having length greater than 5 into lowercase using List Comprehension.
2. Print a specified list after removing the 0th, 4th, and 5th elements.

```python
# Lab Task 05 - Topic: List Comprehensions

# Part (i): List Comprehension Filtering & Lowercase
print("===== Part (i): List Comprehension =====")
words = ["Pakistan", "Apple", "PYTHON", "Karachi", "AI", "Computer", "Hello"]

print("Original List:")
print(words)

# Convert strings having length > 5 into lowercase
lower_words = [word.lower() for word in words if len(word) > 5]

print("\nLowercase Strings (Length > 5):")
print(lower_words)

# Part (ii): Remove 0th, 4th and 5th Elements
print("\n===== Part (ii): Remove Elements =====")
colors = ['Red', 'Green', 'White', 'Black', 'Pink', 'Yellow', 'Teapink']

print("Original List:")
print(colors)

# Keep only elements except index 0, 4 and 5
new_list = [colors[i] for i in range(len(colors)) if i not in (0, 4, 5)]

print("\nUpdated List:")
print(new_list)
```
